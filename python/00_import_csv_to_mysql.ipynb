### **Import pandas**

```python
import pandas as pd
from sqlalchemy import create_engine

password = input('MySQL password: ')
engine = create_engine(f'mysql+mysqlconnector://root:{password}@127.0.0.1:3306/projektas')
```

## **Pievos ir ganyklos**

```python
for year in range(2020, 2026):
    for chunk in pd.read_csv(f'PievosGanyklos/pievosirganyklos_{year}-07-15.csv', encoding='utf-8-sig', sep=';', chunksize=100000):
        chunk.to_sql(f'pievosirganyklos{year}', con=engine, if_exists='append', index=False)
        print('chunk importuotas ✅')

for chunk in pd.read_csv('PievosGanyklos/pievosirganyklos_2010_2019.csv', encoding='utf-8-sig', sep=';', decimal = ',', chunksize=50000): ## decimal = ',' reikalingas, nes 2010-2019 atskirti visur per kablelį, 2020-2025 -- per tašką
        chunk.to_sql('pievosirganyklos_2010_2019', con=engine, if_exists='append', index=False)
        print('chunk importuotas ✅')
```
